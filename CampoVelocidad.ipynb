$$ \vec{V} = (u,v) = (0.5 + 0.8x)\hat{i} + (1.5-0.8y)\hat{j} $$

```python
#Librerías
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Datos de entrada
dx  = 0.01
dy  = 0.01

x = np.arange(-3,3+dx,dx)
y = np.arange(-1,5.5+dy,dy)

X,Y = np.meshgrid(x,y)

plt.plot(X,Y,'ok')
plt.grid()
plt.xlabel('x')
plt.ylabel('y')
plt.title('Grilla de Coordenadas')
plt.show()
```

```python
#Calculamos el campo de Velocidad
u = (0.5 + 0.8*X)
v = (1.5 - 0.8*Y)

#magnitud de la velocidad
V = np.sqrt(u**2 + v**2)

fig, ax = plt.subplots(1,3,figsize=(15,5))

#Mapas de Colores
ax[0].pcolormesh(X,Y,u,vmin=-3,vmax=3,cmap='jet')
ax[1].pcolormesh(X,Y,v,vmin=-3,vmax=3,cmap='jet')
P =ax[2].pcolormesh(X,Y,V,vmin=-3,vmax=3,cmap='jet')
cbar = fig.colorbar(P)
cbar.ax.set_ylabel('Velocidad (m/s)')
ax[0].set_xlabel('x (m)')
ax[1].set_xlabel('x (m)')
ax[0].set_ylabel('y (m)')
ax[0].set_title('Velocidad Horizontal $u(x) = 0.5 + 0.8*x$')
ax[1].set_title('Velocidad Vertical $v(y) = 1.5 - 0.8*y$')
ax[2].set_title('Magnitud del campo de Velocidad  $|V|$')
plt.show()
```

```python
fig, ax = plt.subplots(1,3,figsize=(15,5))

Q = ax[0].quiver(X[::50,::50],Y[::50,::50],u[::50,::50],v[::50,::50])
plt.quiverkey(Q,0.95,1,10,'10 m/s')
ax[0].set_xlabel('x (m)')
ax[0].set_ylabel('y (m)')
ax[0].grid()
ax[0].set_title('Campo de Velocidad')

C = np.linspace(-15, 15, 50)
xx = np.arange(-3,3+0.1,0.1)

for i in range(len(C)):
  yy = ((C[i])/(0.5+(0.8*xx)))+1.875
  ax[0].plot(xx,yy,'b')

ax[0].set_xlim(-3,3)
ax[0].set_ylim(-1,5.5)

#ax[0].streamplot(X,Y,u,v)#, density=[0.5, 1])

C = ax[1].contour(X,Y,V,colors='k')
ax[1].set_xlabel('x (m)')
ax[1].set_ylabel('y (m)')
ax[1].grid()
ax[1].set_title('Mapa de Isotacas')

CF = ax[2].contourf(X,Y,V,cmap='jet')
ax[2].set_xlabel('x (m)')
ax[2].set_ylabel('y (m)')
ax[2].grid()
ax[2].set_title('Mapa de Contornos')

cbar = fig.colorbar(CF)
cbar.ax.set_ylabel('Velocidad (m/s)')

plt.clabel(C,fontsize=10)
plt.show()
```
