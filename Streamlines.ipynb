Campo 2D de Velocidad:

$$\vec{V}=(u,v)=(0.5+0.8x)\hat{i}+(1.5+2.5\sin(2\pi t)-0.8y)\hat{j}$$

## Líneas de Corriente

Las líneas de corriente en el plano $xy$ se determinan como:

$$\left(\frac{dy}{dx} \right)_{a \ lo \ largo \ de \ una \ línea \ de \ corriente}=\frac{v}{u}$$

Con el campo de velocidad:

$$\vec{V}=(u,v)=(0.5+0.8x)\hat{i}+(1.5+2.5\sin(2\pi t)-0.8y)\hat{j}$$

Sustituyendo $u$ y $v$:

$$\frac{dy}{dx}=\frac{1.5+2.5\sin(2\pi t)-0.8y}{0.5+0.8x}$$

Definimos:

$$A = 1.5+2.5\sin(2\pi t)$$

Luego:

$$\frac{dy}{dx} =\frac{A-0.8y}{0.5+0.8x} $$

Transformamos variables, definiendo $X$ e $Y$:

$$0.5 + 0.8x = 0.8(x + 0.625) = 0.8 X \qquad ⟹ X = x + 0.625$$

$$ Y = y - \frac{A}{0.8} $$

La EDO toma la forma:

$$ \frac{dY}{dX} = - \frac{Y}{X} $$

Separando variables:

$$ \frac{dY}{Y} = - \frac{dX}{X} $$

Integrando y aplicando propiedades de Logaritmos:

$$ \int\frac{dY}{Y} = - \int\frac{dX}{X}$$

$$ \ln|Y| = - \ln|X| + C$$

$$ \ln|XY| = C$$

Por lo tanto:

$$ XY = C_{1}$$

Regresando a las variables originales:

$$ (x + 0.625)\left( y - \frac{A}{0.8} \right) = C_{1}$$

Reordenando y recordando que $A = 1.5+2.5\sin(2\pi t)$:

$$ y = \frac{C_{1}}{(x + 0.625)} + \frac{1.5+2.5\sin(2\pi t)}{0.8} $$

Esta es la familia de líneas de corriente. Cada valor de $C_{1}$ representa una línea de corriente diferente.

Si consideramos que la línea de corriente que pasa por $(x,y) = (0.5,0.5)$ en el tiempo $t=0$, se tiene:

$$ y = \frac{C_{1}}{(x + 0.625)} + \frac{1.5}{0.8} ⟹ 0.5 = \frac{C_{1}}{(0.5 + 0.625)} + \frac{1.5}{0.8}$$

$$ C_{1} = 0.5(0.5 + 0.625)-\frac{1.5}{0.8} = -1.3125 $$

Así la línea de corriente que pasa por $(x,y) = (0.5,0.5)$ en el tiempo $t=0$ es:

$$ y = \frac{-1.3125}{(x + 0.625)} + \frac{1.5+2.5\sin(2\pi t)}{0.8} $$


## Líneas de Trayectoria

Ahora queremos saber cuál es el camino que sigue una partícula específica. Por definición:

\begin{align}
u = \frac{dx}{dt} &\qquad\qquad v = \frac{dy}{dt}\\
0.5+0.8x = \frac{dx}{dt} &\qquad\qquad 1.5+2.5\sin(2\pi t)-0.8y &= \frac{dy}{dt}
\end{align}

Por lo tanto:

$$\frac{dx}{dt} = 0.5+0.8x$$

$$\frac{dy}{dt} = 1.5+2.5\sin(2\pi t)-0.8y$$

Supongamos que la partícula inicialmente se encuentra en:

$$ x(0) = x_{0}$$
$$ y(0) = y_{0}$$

Comenzando por la primera EDO:
\begin{align}
\frac{dx}{dt}    &= 0.5+0.8x \quad (\text{tomamos } X = x + 0.625) \\
\frac{dX}{dt} &= 0.8X \\
\frac{dX}{X} &= 0.8 dt \\
\int\frac{dX}{X} &= 0.8\int dt \\
\ln|X| &= 0.8t + C \\
X &= e^{0.8t+C} \\
X &= C_{1}e^{0.8t} \\
x + 0.625 &= C_{1}e^{0.8t} \\
x(t) &= C_{1}e^{0.8t} -0.625
\end{align}

Aplicando $x(0) = x_{0}$, se tiene:

$$ C_{1} = x_{0} + 0.625 $$

Finalmente:
$$ x(t) = (x_{0} + 0.625)e^{0.8t} -0.625  $$

Para la segunda EDO:
\begin{align}
\frac{dy}{dt} &= 1.5+2.5\sin(2\pi t)-0.8y \\
y' + 0.8 y &= 1.5+2.5\sin(2\pi t)
\end{align}

Una forma sencilla de resolver es buscar una solución particular:

$$ y_{p} = A\sin(2\pi t) + B\cos(2\pi t) + C $$

Cuya derivada es:

$$ y_{p}' = 2\pi A\cos(2\pi t) - 2\pi B \sin(2\pi t)$$

Reemplando las dos expresiones anteriores en lado izquierdo de la EDO, se tiene:

\begin{align}
y' + 0.8y & = 2\pi A\cos(2\pi t) - 2\pi B \sin(2\pi t) + 0.8A\sin(2\pi t) + 0.8B\cos(2\pi t) + 0.8C  \\
 &= (2\pi A + 0.8B)\cos(2\pi t) + (0.8A - 2\pi B)\sin(2\pi t) + 0.8C
\end{align}

Esto debe ser igual a $1.5+2.5\sin(2\pi t)$. Por comparación, se tiene:
\begin{align}
2\pi A +0.8B & = 0  \\
0.8A - 2\pi B &= 2.5 \\
0.8C = 1.5
\end{align}

Se tiene un sistema con 3 ecuaciones y 3 incógnitas. La solución es:  
$$C = 1.5/0.8 = 1.875$$

$$ B = -\frac{2\pi}{0.8}A \qquad \text{sustituyendo en } 0.8A + \frac{(2\pi)^2}{0.8}A$$

Por lo tanto:

$$A = \frac{2}{0.64 + 4\pi^2} $$

y:

$$ B = - \frac{5\pi}{0.64 + 4\pi^2} $$

La solución general es:
$$ y(t) = y_{p} + C_{1}e^{-0.8t} = A\sin(2\pi t) + B\cos(2\pi t) + C + C_{1}e^{-0.8t} $$

Aplicando $y(0) = y_{0}$, se tiene:

$$ y_{0} = C + B + C_{1}$$

Así:

$$C_{1} = y_{0} - C - B = y_{0} - 1.875 + \frac{5\pi}{0.64 + 4\pi^2}$$

Por lo tanto:

$$ y(t) = A\sin(2\pi t) + B\cos(2\pi t) + C + \left[y_{0} - 1.875 + \frac{5\pi}{0.64 + 4\pi^2}\right]e^{-0.8t}  $$

Escribiendo todo explícitamente, se tiene:
$$ y(t) = 1.875 + 0.04985\sin(2\pi t) - 0.39129\cos(2\pi t) + \left[y_{0} - 1.48371\right]e^{-0.8t}  $$

```python
#Datos de Entrada

dt      = 0.05 #paso de tiempo (sec)
t_total = 2.0  #tiempo total de la animación (sec)

#Puntos (x0,y0) del origen de la inyección y del origen de la linea de trayectoria
x0 = [0.5, 0.5, 0.5]
y0 = [0.5, 2.5, 4.5]

#tamaño de espaciamiento en el espacio
dx = 0.05
dy = 0.05

#rango de valores 2D
xmin = 0
xmax = 6
ymin = 0
ymax = 6

#-----------------------------------------
#-----------------------------------------
nt = int(t_total/dt)
XT = []
YT = []
XT1 = []
YT1 = []
XT2 = []
YT2 = []
XH = []
YH = []

x = np.arange(xmin,xmax+dx,dx)
y = np.arange(ymin,ymax+dy,dy)

#Creamos el grafico
alto  = 5
ancho = 5

fig1, ax = plt.subplots(figsize=(ancho,alto))

#Definir cosas que no se mueven
ax.set_xlabel('x (m)',fontsize=12)
ax.set_ylabel('y (m)',fontsize=12)
ax.set_xlim(xmin,xmax)
ax.set_ylim(ymin,ymax)

#Lineas de Trayectoria (predibujadas)
tt = np.arange(0,t_total+dt,dt)
for i in range(len(x0)):
  xxt = (x0[i] + 0.625)*np.exp(0.8*tt) - 0.625
  yyt = 1.875 + 0.04985*np.sin(2*np.pi*tt) - 0.39129*np.cos(2*np.pi*tt) + (y0[i] - 1.48371)*np.exp(-0.8*tt)
  ax.plot(xxt,yyt,linewidth=0.25,linestyle='--', color='r',zorder=2)
  if i == 0:
    ax.scatter(x0[i],y0[i], facecolors='none', edgecolors='orange', s=25, linewidths=1.5,label='puntos de inyección')
  else:
    ax.scatter(x0[i],y0[i], facecolors='none', edgecolors='orange', s=25, linewidths=1.5)

X, Y = np.meshgrid(x,y)

u = 0.5 + 0.8*X
v = 1.5 + 2.5*np.sin(2*np.pi*0) - 0.8*Y

nq = 10

#mapa de colores
#quad = ax.pcolormesh(X,Y,np.zeros(np.shape(X)),shading='gouraud',cmap=plt.cm.jet, vmin=0, vmax = 5)
#mapa de flechas
Q = ax.quiver(X[::nq,::nq], Y[::nq,::nq], u[::nq,::nq], v[::nq,::nq], pivot='tail', color='k')
#escala de flechas
plt.quiverkey(Q,0.95,1.03,6.5,'6.5 = m/s')#, labelpos="E")

#eje de lineas de Trayectoria
line, = ax.plot([],[],linewidth=1.0, color='r',zorder=2,label='Líneas de Trayectoria')
lineT1, = ax.plot([],[],linewidth=1.0, color='r',zorder=2)
lineT2, = ax.plot([],[],linewidth=1.0, color='r',zorder=2)
#Scatter para Posición de Trayectoria de una Partícula de Fluido
lineb = ax.scatter([], [], s=25, c='r', edgecolors='k', linewidths=0.5,zorder=50,label = 'Trayectoria de Partícula')
linebT1 = ax.scatter([], [], s=25, c='r', edgecolors='k', linewidths=0.5,zorder=50)
linebT2 = ax.scatter([], [], s=25, c='r', edgecolors='k', linewidths=0.5,zorder=50)
#eje de lineas de Corrientes
line2, = ax.plot([],[],linewidth=1.0, color='b',zorder=2,label='Líneas de Corriente')
line2T1, = ax.plot([],[],linewidth=1.0, color='b',zorder=2)
line2T2, = ax.plot([],[],linewidth=1.0, color='b',zorder=2)
#eje de lineas de Humo
line3, = ax.plot([],[],linewidth=1.0, color='g',zorder=2,label='Líneas de Humo')
line3T1, = ax.plot([],[],linewidth=1.0, color='g',zorder=2)
line3T2, = ax.plot([],[],linewidth=1.0, color='g',zorder=2)
#Scatter para Posición de varias Partícula de Fluido que pasaron por el punto (xs,ys)
linec = ax.scatter([], [], s=25, c='darkblue', linewidths=0.5,zorder=10,label = 'Partículas que pasaron por $(x_{0},y_{0})$')
linecT1 = ax.scatter([], [], s=25, c='darkblue', linewidths=0.5,zorder=10)
linecT2 = ax.scatter([], [], s=25, c='darkblue', linewidths=0.5,zorder=10)

#ax.legend(loc='upper right',framealpha=1.0,fontsize=8)
handles, labels = plt.gca().get_legend_handles_labels()
order = [0,2,5,1,3,4]
ax.legend([handles[idx] for idx in order],[labels[idx] for idx in order],loc='upper right',framealpha=1.0,fontsize=8)

#paleta para el mapa de colores
#cbar = fig1.colorbar(quad)
#cbar.ax.set_ylabel('Velocidad (m/s)',fontsize=15)

#titulo (vacío, pero será dinámico)
time_text = ax.set_title('',fontsize=12)
```

```python
XT = []
YT = []
XT1 = []
YT1 = []
XT2 = []
YT2 = []
XH = []
YH = []

def pathlines(x0,y0,t):
  xt = (x0 + 0.625)*np.exp(0.8*t) - 0.625
  yt = 1.875 + 0.04985*np.sin(2*np.pi*t) - 0.39129*np.cos(2*np.pi*t) + (y0 - 1.48371)*np.exp(-0.8*t)
  return xt, yt

def streamlines(x,xlc,ylc,t):
  C1 = (xlc + 0.625)*(ylc - ((1.5 + 2.5*np.sin(2*np.pi*t))/0.8))
  y = (C1/(x + 0.625)) + ((1.5+2.5*np.sin(2*np.pi*t))/0.8)
  return x,y

def streaklines(xs,ys,dt,t):
  tau = np.arange(0,t+dt,dt)
  xh = (xs + 0.625)*np.exp(0.8*(t-tau)) - 0.625
  yh = 1.875 + 0.04985*np.sin(2*np.pi*t) - 0.39129*np.cos(2*np.pi*t) + (ys - 1.875 - 0.04985*np.sin(2*np.pi*tau) + 0.39129*np.cos(2*np.pi*tau))*np.exp(-0.8*(t-tau))
  return xh, yh


def animate_0(t):
  t = t*dt
  #Campo de velocidad
  u = 0.5 + 0.8*X
  v = 1.5 + 2.5*np.sin(2*np.pi*(t)) - 0.8*Y
  #Magnitud del vector velocidad
  V = np.sqrt(u**2 + v**2)
  #Lineas de Trayectoria
  xt, yt = pathlines(x0[0],y0[0],t)
  #print(xt)
  #print(yt)
  XT.append(xt)
  YT.append(yt)

  xt1, yt1 = pathlines(x0[1],y0[1],t)
  XT1.append(xt1)
  YT1.append(yt1)

  xt2, yt2 = pathlines(x0[2],y0[2],t)
  XT2.append(xt2)
  YT2.append(yt2)

  #Lineas de corriente
  xc, yc = streamlines(x,xt,yt,t)
  xc1, yc1 = streamlines(x,xt1,yt1,t)
  xc2, yc2 = streamlines(x,xt2,yt2,t)
  #lineas de Humo
  xh,yh = streaklines(x0[0],y0[0],dt,t)
  xh1,yh1 = streaklines(x0[1],y0[1],dt,t)
  xh2,yh2 = streaklines(x0[2],y0[2],dt,t)

  if t == 0:
    time_text.set_text('Tiempo = %0.1f (s)' % (t,))
    #XH.append(xh[-1])
    #YH.append(yh[-1])
  elif  t >0 and int(t/dt) % 2 == 0:
    time_text.set_text('Tiempo = %0.1f (s)' % (t,))
    #XH.append(xh[-1])
    #YH.append(yh[-1])

  #time_text.set_text('Tiempo = %0.2f (s)' % (t,))
  #quad.set_array(V.ravel())
  Q.set_UVC(u[::nq,::nq],v[::nq,::nq])
  line.set_data(XT,YT)
  lineT1.set_data(XT1,YT1)
  lineT2.set_data(XT2,YT2)
  lineb.set_offsets([xt,yt])
  linebT1.set_offsets([xt1,yt1])
  linebT2.set_offsets([xt2,yt2])
  line2.set_data(xc,yc)
  line2T1.set_data(xc1,yc1)
  line2T2.set_data(xc2,yc2)
  line3.set_data(xh,yh)
  line3T1.set_data(xh1,yh1)
  line3T2.set_data(xh2,yh2)
  #vector = np.c_[np.array([XH, YH])]
  #linec.set_offsets(vector)
  linec.set_offsets(np.c_[xh[::5],yh[::5]])
  linecT1.set_offsets(np.c_[xh1[::5],yh1[::5]])
  linecT2.set_offsets(np.c_[xh2[::5],yh2[::5]])
  return t,#Q#, time_text

anim = animation.FuncAnimation(fig1, animate_0, frames=nt+1, interval=100, blit=False)

#anim.save(filename='example.mp4', writer='ffmpeg',bitrate=20)

HTML(anim.to_jshtml())
```
